# Abstract 
Our goal is to compare different algorithms that can be used to solve Connect 6. This game features a 2 player 19x19 board where the first player to make a chain of 6 or more vertically, horizontally or diagonally wins the game. We will compare different algorithms such as minimax with alpha beta pruning, reinforcement learning, greedy heuristic search etc. We plan to make our own data which will be composed of different Connect 6 games. Specifically, this data will be generated through simulated gameplay where each game will have board configurations, possible moves, rewards and outcomes. Each board state will contain black and white positions and the current player’s turn.  This is relevant mostly to reinforcement learning algorithms as they require hyperparameter tuning. We will compare the different algorithms on the following quantitative metrics: win rate (%), average decision time to make a single move (in sec) and average game length (number of turns it takes to finish the game).

# Problem Statement
In this project, we will aim to balance strategic depth and computational efficiency in evaluating search algorithms to strategically choose the optimal action in the game of *Connect-Six* (a variant of Gomoku with a six-stone victory condition) by using action-space reduction techniques to mitigate combinatorial complexity. We will evaluate MCTS policy networks and Reinforcement Learning technique against a baseline of Alpha-Beta Minimax evaluation, utilizing a standard 19x19 Connect-Six Board with standard rules (First move is 1 stone, all other moves are alternated with 2 stones per move until one player gets 6 stones in a row, vertically, horizontally or diagonally).

# Data

The dataset will be generated through simulated Connect 6 games using a game environment that implements the standard Connect 6 rules. Different AI agents that we  implemented will interact with this environment by observing the current board and selecting moves. The environment will validate moves, update the board, determine winning conditions, and record the gameplay. We aim to include at least 2000 games. Each observation will represent a board state at the beginning of a player's turn. The board will be represented using 361 variables corresponding to its 19 × 19 cells, encoded as 1 (Black stone), -1 (White stone), or 0 (empty). Other variables record the current player, selected moves, rewards, and game outcome. The resulting records will be stored in CSV or JSON format and used to train reinforcement learning agents. 

# Proposed Solution

In this project, we will compare a set of search algorithms to a benchmark model using depth-limited Minimax with Alpha-Beta pruning and a heuristic evaluation function. This is a classical deterministic search algorithm that performs with reasonable success at Connect-six and the Gomoku class of games. Our proposed solution would have to construct an adversial search tree while mitigating the large branching factor of each state. To that end, we will evaluate the performance of the following 2 algorithms in comparison to the benchmark model:
1. Enhanced AlphaZero with Monte Carlo Tree Search: By combining MCTS with Reinforcement Learning, AlphaZero would have convolutional layers to preserve the grid topology in its learning. Furthermore, it would ideally develop dynamic knowledge to select stones that are not independently placed but rather used strategically. We can also set exploration/exploitation bounds to make this solution effective and efficient. In our implementation, the policy network evaluates the first stone ($a_1$) across 361 outputs, concatenates a spatial mask of $a_1$ onto the feature map, and then evaluates the second stone ($a_2$) across 360 outputs. This reduces output dimension from $64,620$ to $721$, making AlphaZero scalable to Connect6.
2. Model-free Proximal Policy Optimization: As a contrast to the model-based MCTS engine above, PPO would train an Actor-Critic network using an advantage estimator. Actor-Critic networks are commonly used in solving games. This solution would ideally calculate the benefit of policy gradient learning in the model in this game.
