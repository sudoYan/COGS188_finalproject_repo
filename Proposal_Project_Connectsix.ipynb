# Abstract 
Our goal is to compare different algorithms that can be used to solve Connect 6. This game features a 2 player 19x19 board where the first player to make a chain of 6 or more vertically, horizontally or diagonally wins the game. We will compare different algorithms such as minimax with alpha beta pruning, reinforcement learning, greedy heuristic search etc. We plan to make our own data which will be composed of different Connect 6 games. Specifically, this data will be generated through simulated gameplay where each game will have board configurations, possible moves, rewards and outcomes. Each board state will contain black and white positions and the current player’s turn.  This is relevant mostly to reinforcement learning algorithms as they require hyperparameter tuning. We will compare the different algorithms on the following quantitative metrics: win rate (%), average decision time to make a single move (in sec) and average game length (number of turns it takes to finish the game). 

# Background

Connect Six is a two-player strategy game introduced by Wu, Huang, and Chang (2005) as a variation of the traditional Gomoku game. Unlike Gomoku, where players place one stone per turn, Connect Six allows players to place two stones per turn after the first move. Although the rules are relatively simple, the game has a large number of possible board configurations and possible moves, making it difficult for computer programs to evaluate every possible outcome. Previous research has also examined the fairness and complexity of Connect Six, highlighting the challenges involved in developing effective strategies for the game [1].

One common approach to solving strategy games is Minimax search, which evaluates possible future moves by considering the decisions of both players. Alpha-Beta pruning improves this approach by removing branches of the search tree that cannot influence the final decision. However, Connect Six presents a particular challenge because of its large number of possible moves. Research by Yen et al. (2012) explored ways to improve the efficiency of Connect Six programs through threat-based move generation, bitboard representations, and specialized search strategies. Their work demonstrates the importance of reducing the number of possible moves that need to be evaluated when developing a competitive Connect Six agent [2].

Another approach to solving strategy games is Monte Carlo Tree Search (MCTS), which combines tree search with simulations to estimate the potential outcomes of different moves. Browne et al. (2012) explain how MCTS has been successfully applied to complex games, particularly Go, by focusing computational resources on promising moves rather than exploring every possible outcome [3]. This approach was further developed by Silver et al. (2018) through AlphaZero, which combines MCTS with deep neural networks and reinforcement learning. AlphaZero learns through self-play, using a policy network to guide move selection and a value network to estimate the likelihood of winning from a given board position. Its success in chess, shogi, and Go demonstrates the potential of combining search algorithms with reinforcement learning [4].

In addition to search-based approaches, reinforcement learning allows an agent to improve its strategy through repeated interactions with an environment. One such method is Proximal Policy Optimization (PPO), introduced by Schulman et al. (2017). PPO uses policy gradient learning to update an agent's decisions while limiting how much its policy changes during each update. This helps make the training process more stable and efficient. PPO has been successfully tested in several environments, including video games and simulated tasks, making it another approach worth investigating for Connect Six [5].

Based on these previous approaches, our project will compare Minimax with Alpha-Beta pruning, AlphaZero with MCTS, and PPO to determine how effectively each algorithm can play Connect Six. We are particularly interested in understanding the trade-off between strategic performance and computational efficiency, since an algorithm that plays well may also require significantly more time to select its moves. By comparing these approaches within the same game environment, we hope to identify which methods provide the best balance between playing strength and computational cost.

# Evaluation Metrics

To compare the performance of the three algorithms, we will use three evaluation metrics: win rate, average decision time, and average game length. These metrics will allow us to evaluate both the strategic performance and computational efficiency of each algorithm.

1. Win Rate (%)

Win rate measures how often an algorithm wins against its opponents. It will be calculated by dividing the total number of games won by the total number of games played and multiplying by 100.

Win Rate = (Number of Games Won / Total Games Played) × 100

A higher win rate indicates that an algorithm is more successful at selecting strategic moves and winning games. We will evaluate each algorithm against the other algorithms and alternate which player moves first to account for any potential advantage from playing first.

2. Average Decision Time (seconds)

Average decision time measures how long an algorithm takes to select its move. It will be calculated by recording the time taken to select an action at each turn and dividing the total decision time by the number of decisions made.

Average Decision Time = Total Time Spent Making Decisions / Total Number of Decisions

This metric will help us understand the computational efficiency of each algorithm. Since AlphaZero and PPO use neural networks, while Minimax relies on search and heuristic evaluation, their decision times may differ significantly. A lower average decision time indicates that an algorithm can select moves more quickly.

3. Average Game Length (turns)

Average game length measures the average number of turns required for a game to reach a final outcome. It will be calculated by dividing the total number of turns across all completed games by the total number of games played.

Average Game Length = Total Number of Turns Across Games / Total Number of Games

This metric will help us understand how the different algorithms approach the game. For example, an algorithm that consistently wins in fewer turns may be better at creating winning opportunities. However, a shorter game does not necessarily indicate better performance, since an algorithm could also lose quickly.

All three algorithms will be evaluated using the same game environment and rules. We will use the same hardware and comparable testing conditions when measuring decision time. Each algorithm will play multiple games against the others, with the starting player alternated to reduce first-player bias. Together, these metrics will allow us to compare how effectively each algorithm plays Connect Six while considering the time and computational resources required to make decisions.

# Ethics & Privacy

Since our project uses simulated Connect Six games, there are no major concerns involving personal information or data privacy. No data from real individuals will be collected or stored. However, we will consider the computational resources required to train and evaluate our algorithms, particularly AlphaZero and PPO, which may require a large number of simulated games and repeated training sessions. We will try to limit unnecessary training and use reasonable hyperparameter settings to avoid excessive computational costs.

Another consideration is ensuring that our results are fair and reliable. Different algorithms may perform differently depending on the starting player or testing conditions. To address this, we will use the same game environment and rules for all three algorithms, alternate the starting player, and evaluate performance across multiple games. We will also acknowledge the limitations of our results, since an algorithm's performance in a simulated Connect Six environment does not necessarily represent its performance in other applications.
